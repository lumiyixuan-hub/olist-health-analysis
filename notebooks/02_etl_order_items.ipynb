# Data Cleaning — Order Items Table
**Source:** `olist_order_items_dataset.csv`  
**Output:** `data/cleaned/order_items_cleaned.csv`

### Cleaning Steps
1. Load raw data
2. Initial inspection (shape, columns, data types, missing values, column. summary)
3. Check composite primary key duplicates
4. Convert shipping_limit_date to datetime
5. Export cleaned data

## Step 1 — Load Raw Data

```python
import pandas as pd

# ── 1. Load raw data ──────────────────────────────────────────
df = pd.read_csv('../data/raw/olist_order_items_dataset.csv')
```

## Step 2 — Initial Inspection

```python
# ── 2. Initial inspection ─────────────────────────────────────
print("=== Shape ===")
print(df.shape)

print("\n=== Column names ===")
print(df.columns.tolist())

print("\n=== First 5 rows ===")
print(df.head())

print("\n=== Current dtypes ===")
print(df.dtypes)

print("\n=== Missing values ===")
print(df.isnull().sum())

print("\n=== Numeric columns summary ===")
print(df.describe())

print("\n=== Categorical columns summary ===")
print(df.describe(include='str'))
```

```python
# ── Check zero freight value ──────────────────────────────────
print(f"Zero freight_value: {len(df[df['freight_value'] == 0])}")
print(f"Price below 1: {len(df[df['price'] < 1])}")
```

## Step 3 — Check Composite Primary Key Duplicates

In this table, no single column uniquely identifies a row.
A single order can contain multiple items, so the primary key is the combination of `order_id` + `order_item_id`.
For example, order A with 3 items would have rows: (A, 1), (A, 2), (A, 3).

We check that this combination has no duplicates to ensure data integrity.

```python
# ── 3. Check composite primary key duplicates ─────────────────
# Primary key = order_id + order_item_id (one order can have multiple items)
duplicate_count = df.duplicated(subset=['order_id', 'order_item_id']).sum()
print(f"Duplicate (order_id + order_item_id): {duplicate_count}")
```

## Step 4 — Convert shipping_limit_date to datetime

```python
# ── 4. Convert shipping_limit_date to datetime ────────────────
df['shipping_limit_date'] = pd.to_datetime(df['shipping_limit_date'])

print("=== Dtype after conversion ===")
print(df['shipping_limit_date'].dtype)
```

## Step 5 — Export Cleaned Data

```python
# ── 5. Export cleaned data ────────────────────────────────────
df.to_csv('../data/cleaned/order_items_cleaned.csv', index=False)

print(f"Exported: {len(df)} rows")
print("Saved to: data/cleaned/order_items_cleaned.csv")
```
