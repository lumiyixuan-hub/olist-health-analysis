# ETL Cleaning — Products Table
**Source:** `olist_products_dataset.csv`  
**Output:** `data/cleaned/products_cleaned.csv`

### Key characteristics of this table
- Contains product-level attributes including category, dimensions, and weight
- Missing values are intentionally retained — deleting rows would cause JOIN failures in downstream analysis

### Cleaning Steps
1. Load raw data
2. Initial inspection (shape, columns, missing values, distributions)
3. Verify primary key
4. Export cleaned data

## Step 1 — Load Raw Data

```python
import pandas as pd

# ── 1. Load raw data ──────────────────────────────────────────
df = pd.read_csv('../data/raw/olist_products_dataset.csv')
```

## Step 2 — Initial Inspection

```python
# ── 2. Initial inspection ─────────────────────────────────────
print("=== Shape ===")
print(df.shape)

print("\n=== Column names ===")
print(df.columns.tolist())

print("\n=== First 5 rows ===")
print(df.head())

print("\n=== Current dtypes ===")
print(df.dtypes)

print("\n=== Missing values ===")
print(df.isnull().sum())

print("\n=== Numeric columns summary ===")
print(df.describe())

print("\n=== Categorical columns summary ===")
print(df.describe(include='str'))
```

## Step 3 — Verify Primary Key

`product_id` should be unique — one row per product.  
Missing values in `product_category_name` and physical dimensions are retained.  
These products exist in `order_items` — removing them would break downstream JOINs.

```python
# ── 3. Verify primary key ─────────────────────────────────────
print(f"Total rows: {len(df)}")
print(f"Unique product_id: {df['product_id'].nunique()}")

# Missing values are retained — these products exist in order_items
# Deleting them would cause JOIN failures in downstream analysis
print(f"\nMissing product_category_name: {df['product_category_name'].isnull().sum()} (retained)")
print(f"Missing physical dimensions: {df['product_weight_g'].isnull().sum()} (retained)")
```

## Step 4 — Export Cleaned Data

```python
# ── 4. Export cleaned data ────────────────────────────────────
df.to_csv('../data/cleaned/products_cleaned.csv', index=False)

print(f"Exported: {len(df)} rows")
print("Saved to: data/cleaned/products_cleaned.csv")
```
