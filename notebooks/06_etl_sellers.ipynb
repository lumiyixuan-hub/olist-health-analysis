# ETL Cleaning — Sellers Table
**Source:** `olist_sellers_dataset.csv`  
**Output:** `data/cleaned/sellers_cleaned.csv`

### Cleaning Steps
1. Load raw data
2. Initial inspection (shape, columns, missing values, distributions)
3. Verify primary key
4. Export cleaned data

## Step 1 — Load Raw Data

```python
import pandas as pd

# ── 1. Load raw data ──────────────────────────────────────────
df = pd.read_csv('../data/raw/olist_sellers_dataset.csv')
```

## Step 2 — Initial Inspection

```python
# ── 2. Initial inspection ─────────────────────────────────────
print("=== Shape ===")
print(df.shape)

print("\n=== Column names ===")
print(df.columns.tolist())

print("\n=== First 5 rows ===")
print(df.head())

print("\n=== Current dtypes ===")
print(df.dtypes)

print("\n=== Missing values ===")
print(df.isnull().sum())

print("\n=== Numeric columns summary ===")
print(df.describe())

print("\n=== Categorical columns summary ===")
print(df.describe(include='str'))
```

## Step 3 — Verify Primary Key

`seller_id` should be unique — one row per seller.

```python
# ── 3. Verify primary key ─────────────────────────────────────
print(f"Total rows: {len(df)}")
print(f"Unique seller_id: {df['seller_id'].nunique()}")
```

## Step 4 — Export Cleaned Data

```python
# ── 4. Export cleaned data ────────────────────────────────────
df.to_csv('../data/cleaned/sellers_cleaned.csv', index=False)

print(f"Exported: {len(df)} rows")
print("Saved to: data/cleaned/sellers_cleaned.csv")
```
