# ETL Cleaning — Category Translation Table
**Source:** `product_category_name_translation.csv`  
**Output:** `data/cleaned/category_translation_cleaned.csv`

### Key characteristics of this table
- Simple lookup table mapping Portuguese category names to English
- Used to join with `products` table to display English category names in analysis

### Cleaning Steps
1. Load raw data
2. Initial inspection
3. Verify primary key and export

## Step 1 — Load Raw Data

```python
import pandas as pd

# ── 1. Load raw data ──────────────────────────────────────────
df = pd.read_csv('../data/raw/product_category_name_translation.csv')
```

## Step 2 — Initial Inspection

```python
# ── 2. Initial inspection ─────────────────────────────────────
print("=== Shape ===")
print(df.shape)

print("\n=== Column names ===")
print(df.columns.tolist())

print("\n=== First 5 rows ===")
print(df.head())

print("\n=== Current dtypes ===")
print(df.dtypes)

print("\n=== Missing values ===")
print(df.isnull().sum())

print("\n=== Categorical columns summary ===")
print(df.describe(include='str'))
```

## Step 3 — Verify Primary Key and Export

`product_category_name` should be unique — one row per category.  
No cleaning required — table is complete and consistent.

```python
# ── 3. Verify primary key & export ───────────────────────────
print(f"Total rows: {len(df)}")
print(f"Unique product_category_name: {df['product_category_name'].nunique()}")

df.to_csv('../data/cleaned/category_translation_cleaned.csv', index=False)
print(f"\nExported: {len(df)} rows")
print("Saved to: data/cleaned/category_translation_cleaned.csv")
```
