# ETL Cleaning — Customers Table
**Source:** `olist_customers_dataset.csv`  
**Output:** `data/cleaned/customers_cleaned.csv`

### Key characteristics of this table
- `customer_id` is a per-order identifier — one person placing multiple orders gets a different `customer_id` each time
- `customer_unique_id` is the true person-level identifier — duplicates are expected and normal

### Cleaning Steps
1. Load raw data
2. Initial inspection (shape, columns, missing values, distributions)
3. Verify primary key
4. Export cleaned data

## Step 1 — Load Raw Data

```python
import pandas as pd

# ── 1. Load raw data ──────────────────────────────────────────
df = pd.read_csv('../data/raw/olist_customers_dataset.csv')
```

## Step 2 — Initial Inspection

```python
# ── 2. Initial inspection ─────────────────────────────────────
print("=== Shape ===")
print(df.shape)

print("\n=== Column names ===")
print(df.columns.tolist())

print("\n=== First 5 rows ===")
print(df.head())

print("\n=== Current dtypes ===")
print(df.dtypes)

print("\n=== Missing values ===")
print(df.isnull().sum())

print("\n=== Numeric columns summary ===")
print(df.describe())

print("\n=== Categorical columns summary ===")
print(df.describe(include='str'))
```

## Step 3 — Verify Primary Key

`customer_id` should be unique (one row per order).
`customer_unique_id` duplicates are expected — the same person can place multiple orders.

```python
# ── 3. Verify primary key ─────────────────────────────────────
print(f"Total rows: {len(df)}")
print(f"Unique customer_id: {df['customer_id'].nunique()}")
print(f"Unique customer_unique_id: {df['customer_unique_id'].nunique()}")
print(f"\nNote: customer_unique_id < total rows is expected —")
print(f"same person can place multiple orders with different customer_id")
```

## Step 4 — Export Cleaned Data

```python
# ── 4. Export cleaned data ────────────────────────────────────
df.to_csv('../data/cleaned/customers_cleaned.csv', index=False)

print(f"Exported: {len(df)} rows")
print("Saved to: data/cleaned/customers_cleaned.csv")
```
